```python
import pandas as pd
import os

file_path = r"C:\Users\saran\guvi project\cart2insights\data\cleaned\order_reviews_cleaned.csv"

reviews = pd.read_csv(file_path)

output_folder = r"C:\Users\saran\guvi project\cart2insights\data\reviews_parts"
os.makedirs(output_folder, exist_ok=True)

chunk_size = 10000

for i in range(0, len(reviews), chunk_size):
    chunk = reviews.iloc[i:i + chunk_size]
    file_number = i // chunk_size + 1

    chunk.to_csv(
        f"{output_folder}/reviews_part_{file_number}.csv",
        index=False
    )

print("Total rows:", len(reviews))
print("Files created:", len(range(0, len(reviews), chunk_size)))
```
